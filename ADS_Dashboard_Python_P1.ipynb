```python
import matplotlib.pyplot as plt
x=[2,4,6,8,10]
y=[3,6,9,12,15]
plt.plot(x,y, color='black', linestyle='dashed', marker='x', markersize='10', markerfacecolor='white')
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

plt.bar(x,y, color='orange')
plt.title("Bar Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

plt.barh(x,y, color='pink', edgecolor='black')
plt.title("Horizontal Bar Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

plt.scatter(x,y, color='green')
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
marks=(78,12,65,45,48,90,88,95,80,75)
plt.hist(marks, bins=6, color='darkgreen', edgecolor='white')
plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

expenses=[2000,50000,13500,46500,25000]
categories=['Food', 'investment', 'Rent', 'home', 'Friend']
plt.pie(expenses, labels=categories, autopct="%1.1f%%")
plt.title("Monthly Expenditure")
plt.show()
```
